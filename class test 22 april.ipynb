Q1. Class Variable vs Instance Variable

```python
class Student:
    school_name = "UPES"

    def __init__ (self,name,age):
        self.name = name 
        self.age = age 

stu1 = Student("Aman",20)
stu2 = Student("Riya",21)

print("School:",stu1.school_name)
print("School:",stu2.school_name)

print("Stu1:",stu1.name,stu1.age)
print("Stu2:",stu2.name,stu2.age)

```

Q2. Instance Method and Class Method

```python
class Employee:
    company = "ABC Ltd"
    
    def __init__ (self,name,salary):
        self.name = name
        self.salary = salary 
    
    def show_details(self):
        print("Name:",self.name,"|salary:",self.salary,"|Company:",Employee.company)
    
    @classmethod
    def change_company(cls,new_name):
        cls.company = new_name

emp1 = Employee("Aman",300000)
emp2 = Employee("Riya",400000)
print("Before change")
emp1.show_details()
emp2.show_details()

Employee.change_company("XYZ Ltd")

print("After change:")
emp1.show_details()
emp2.show_details()
```

Q3. Static Method

```python
class Mathhelper:
    @staticmethod
    def is_even(num):
        return num % 2 == 0 
    
print(Mathhelper.is_even(8))
print(Mathhelper.is_even(11))

```

Q4. Public and Private Data Members

```python
class BankAccount:
    def __init__ (self,name,balance):
        self.name = name 
        self.__balance = balance 

    def deposit(self,amount):
        self.__balance += amount

    def withdraw(self,amount):
        if amount <= self.__balance:
            self.__balance -= amount
        else:
            print("insufficient balance")
    def show_balance(self):   
        print("Balance:",self.__balance)

acc = BankAccount("Aman",1000)
acc.deposit(5000)
acc.show_balance()

acc.withdraw(3000)
acc.show_balance()

acc.withdraw(1000)
acc.show_balance()
```

Q6. Magic Methods / Operator Overloading

```python
class Product:
    def __init__(self,name,price):
        self.name = name 
        self.price = price
    def __str__(self):
        return f"Product: {self.name}, Price: {self.price}"
    
    def __add__(self,other):
        return self.price + other.price
    
    def _eq__(self,other):
        return self.price == other.price
    
p1 = Product("Laptop",50000)
p2 = Product("Phone",300000)
p3 = Product("Tablet",50000)

print(p1)
print(p1 + p2)
print(p1 == p3)
print(p1 == p2)
```
