```python
a = 4567
r = a
rev = 0

while r > 0:
    digit = a % 10
    rev = rev * 10 + digit

    r = a // 10
    a = r
    

print("Reversed number:", rev)

if rev == 4567:
    print("It is a Palindrome")
else:
    print("It is not a Palindrome")
```
