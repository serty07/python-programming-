4. Reverse Integer

```python
def reverse_integer(n):
    sign = -1 if n < 0 else 1
    n = abs(n)

    rev = 0 
    while n > 0 :
        digit = n % 10 
        rev = rev * 10 + digit 
        n = n // 10 
    return sign * rev 
num = int(input("Enter the number"))
print ("The reversed number",reverse_integer(num))
```

Q1. Palindrome numbers 

```python
num = int(input("Enter the numbers"))
rev = 0 
temp = num 

while temp > 0 :
    rev = rev * 10 + temp % 10 
    temp //= 10 

if num == rev :
    print ( "Palindrome number")
else:
    print("not a palindrome")
```

Q2. fibonacci series 

```python
n = int(input("Enter the number "))

a,b = 1,2
print(a,b,end="")

for i in range (n -2):
    c = a + b 
    print(c,end = "")
    a,b = b,c
```

Q1. Menu-Driven Calculator

```python
a = int(input("Enter the first number"))
b = int(input("Enter the second number"))

print("Choose an operation:")
print("1. Addition")
print("2. Subtraction")
print("3. multiplication")
print("4. Division")
print("5. Modulus ")
print("6. Floor Division")

choice = int(input("Enter your choice"))

if  choice == 1:
    print("Result",a + b)
elif choice == 2:
    print("Result", a - b)
elif choice == 3:
    print("Result", a * b)
elif choice == 4:
    if b != 0:
        print("Result", a / b)
    else :
        print("Error: Division by zero")
elif choice == 5:
    if b != 0:
        print("Result", a % b )
    else:
        print("Error: modulus by zero ")
elif choice == 6 :
    if b != 0:
        print("Result", a // b)
    else:
        print("Error: Floor division by zero")
else:
    print("Invalid choice")
```

* Number guessing Game 

```python
import random 

secret = random.randint(1, 10)
attempts = 0 

print ("Guess the numbers between 1 and 10 ")

while True:
    guess = int(input("Enter your guess"))
    attempts = attempts + 1

    if guess == secret:
        print("Correct")
        print("you guessed it in",attempts,"attempts")
        break
    elif guess > secret:
        print("Too high")
    else: 
        print("too low")

    
```

* Password strength checker 

```python
password = input("Enter the password:")

if len(password) < 6:
    print("Weak Password")
elif password.isdigit() or password.isalpha():
    print("Medium Password")
else:
    print("Strong Password")
    
```

* Word & Character Counter

```python
text = input("Enter a sentence:")

words = text.split()
word_count = len(words)

char_count = 0
for ch in text:
    if ch != " ":
        char_count = char_count + 1
print("Words:",word_count)
print("Characters (without spaces)",char_count)
```

* Stopwatch 


```python
second = 0 

print("Stopwatch Started")
print("Press enter to move time forward")
print("Type 'stop' to end\n")

while True:
    command = input(">>")
    
    if command.lower() == "stop":
        break 

    second += 1

    mins = second // 60
    secs = second % 60
    print(f"time: {mins :02d}:{secs :02d}")

print("\n Stopwatch stopped")
print(f"Final time: {mins :02d}:{secs :02d}")
```
