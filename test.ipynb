```python
def moving_average(stream, w):
    result = []
    window_sum = 0 

    for i in range (len(stream)):
        window_sum += stream[i]

        if i >= w:
            window_sum -= stream[i-w]
    
        window_print = min(i + 1, w)
   
        result.append(window_sum/window_print)
    
    return result

stream = [1,2,3,4,5]
w = 3

print(moving_average(stream, w))
```

```python
def count_by_category(pairs):
    d = {}
    for cat, word in pairs:
        d.setdefault(cat, {})
        d[cat].setdefault(word, 0)
        d[cat][word] += 1
    return d

pairs = [("sports","win"), ("sports","win"), ("tech","ai")]
```

```python
count_by_category(pairs)
```

```python
def missing_number(nums):
    n = len(nums)
    total = n * (n + 1) / 2
    return int(total - sum(nums))
nums = [3, 0, 1]
```

```python
missing_number(nums)
```

```python
def add_unique(x, seen=set()):
    if x not in seen:
        seen.add(x)
    return list(seen)

add_unique(1)
add_unique(2)
```

```python
def top_k(items, k):
    items.sort(key=lambda x: x[1], reverse=True)
    return items[:k]
```

```python
items = [("A", 90), ("B", 95), ("C", 95), ("D", 85)]
k = 2
```

```python
top_k(items, k)
```
