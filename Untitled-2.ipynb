```python
def moving_average(stream, w):
    result = []
    window_sum = 0 

for i in range (len(stream)):
    window_sum += stream(i)

    if i >= w:
        window_sum -= stream[i-w]
    
    window_print = min(i + 1, w)
   
    result.append(window_sum/window_print)
    
return result

stream = [1,2,3,4,5]
w = 3

print(moving_average(stream,w))
```
